# Line flow analysis

```julia
line_flows = retrieve_line_flows(run_dir, result_file, absolute = true, relative = true);
```

```julia
minimum(line_flows), maximum(line_flows)
```

```julia
line_ids = get_ordered_line_ids(network);
```

```julia
function line_info(i)
    id = line_ids[i]
    line = network["branch"][id]
    f_bus = network["bus"][string(line["f_bus"])]
    t_bus = network["bus"][string(line["t_bus"])]
    return "$id : $(f_bus["name"]) - $(t_bus["name"])"
end
```

### Maximum relative flow

```julia
max_flow_by_t = dropdims(maximum(line_flows, dims=1), dims=1);
```

```julia
plot(max_flow_by_t, label="", xaxis="t [hour]", yaxis="Maximum line flow (relative)", legend=:bottomright,
    xlims=(0,8736), ylims=(0,2))
```

```julia
max_flow_by_line = dropdims(maximum(line_flows, dims=2), dims=2);
```

```julia
stephist(max_flow_by_line, label="", xaxis= "Maximum line flow (relative)", yaxis="Number of lines")
```

```julia
(highest_max_val, highest_max_t) = findmax(max_flow_by_t)
highest_max_line = findmax(max_flow_by_line)[2]
highest_max_val, highest_max_line, highest_max_t
```

```julia
plot(line_flows[highest_max_line, :], label=line_info(highest_max_line), xaxis="t [hour]", yaxis="Line flow (relative)",
    legend=:bottomright, xlims=(0,8736), ylims=(0,2))
```

```julia
(lowest_max_val, lowest_max_t) = findmin(max_flow_by_t)
lowest_max_val, lowest_max_t
```

### Average flow

```julia
avg_flow_by_t = dropdims(mean(line_flows, dims=1), dims=1);
```

```julia
plot(avg_flow_by_t, label="", xaxis="t [hour]", yaxis="Average line flow (relative)", legend=:bottomright,
    xlims=(0,8736), ylims=(0,1))
```

```julia
avg_flow_by_line = dropdims(mean(line_flows, dims=2), dims=2);
```

```julia
stephist(avg_flow_by_line, label="", xaxis="Average flow (relative)", yaxis="Number of lines")
```

```julia
(highest_avg_val, highest_avg_line) = findmax(avg_flow_by_line)
```

```julia
plot(line_flows[highest_avg_line, :], label=line_info(highest_avg_line), xaxis="t [hour]", yaxis="Line flow (relative)",
    legend=:bottomright, xlims=(0,8736), ylims=(0,2))
```

### Identify most problematic lines

```julia
max_flow_dict = Dict(line_ids .=> max_flow_by_line)
sort(max_flow_dict, byvalue=true, rev=true)
```

```julia
problematic_lines = [findfirst(==(id), line_ids) for (id, max_flow) in max_flow_dict if max_flow > 1.4]
line_info.(problematic_lines)
```

# Examples of generator outputs

```julia
gen_ids = get_ordered_gen_ids(network)

function plot_gen(gen_id, gen_name)
    i = findfirst(==(gen_id), gen_ids)
    plot(100 * gens[:, i], label="", title=gen_name, xlabel = "Time [h]", ylabel = "Power [MW]",
        xlims = (0, 24*364), ylims = (0, :auto))
end
```

```julia
function list_gens_by_type(gen_type)
    sort([(id, network["bus"][string(gen["gen_bus"])]["name"], 100 * gen["pmax"], gen["type"])
        for (id, gen) ∈ network["gen"] if gen["aggregated_type"] == gen_type], by=val->val[3], rev=true)
end
```

```julia
list_gens_by_type("coal")
```

```julia
plot_gen("325", "Bełchatów (PL) - Brown Lignite")
```

```julia
list_gens_by_type("hydro_storage")
```

```julia
plot_gen("914", "Bieudron/Nendaz (CH) - Hydro")
```

```julia
list_gens_by_type("gas")
```

# Export model

```julia
function export_model(t, label)
    update_network!(network, collect(loads[t, :]), collect(gens[t, :]), collect(lines[t, :]))
    filename = "data/europe_$label.json"
    open(filename, "w") do f
        JSON.print(f, network) 
    end
    filename
end
```
